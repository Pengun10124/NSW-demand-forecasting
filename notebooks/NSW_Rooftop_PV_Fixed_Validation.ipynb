# Rooftop PV experiment: fixed validation dates

Run from the repository root or notebooks directory. This version preserves the original notebook and earlier results. It compares the same Ridge model with no PV, lagged PV only, and observed PV plus history. The joined PV dataset is saved in data/pv; reruns do not need the Desktop source archives.

Training ends March 31, 2026 (990 days). The original five validation windows are fixed. April-June and July-August tests have already been inspected. Same-day observed PV is an explanatory input and would require a forecast in operational use; observed weather is also idealized. No tuning was added.


```python
from pathlib import Path
import runpy
repo = next(p.resolve() for p in [Path.cwd(), Path.cwd().parent] if (p / "experiments/run_pv_experiment.py").is_file())
result = runpy.run_path(str(repo / "experiments/run_pv_experiment.py"))
```

## Results
Lower MAE is better. Historical PV uses lag-1 and lag-7 mean, maximum and midday mean. The observed scenario adds current-day summaries and midday change. All variants use identical training days, preprocessing and Ridge alpha=10.


```python
result["metrics"].pivot_table(index=["Target", "Period"], columns="Variant", values="MAE_MW", sort=False).round(2)
```

## Interpretation of the saved run
Observed PV reduces August errors, but July and the July-August aggregate worsen. Historical PV does not improve the July-August tests either. Validation gains do not establish improvement on the inspected test periods. This experiment is retained for comparison, not promoted as the best model. Missing PV stays blank; incomplete daily PV summaries use the existing training-fitted median imputer. No large demand-error days were removed.
